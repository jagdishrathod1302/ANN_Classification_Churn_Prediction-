```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent=[  'the glass of milk',
     'the glass of juice',
     'the cup of tea',
    'I am a good boy',
     'I am a good developer',
     'understand the meaning of words',
     'your videos are good',]
```

```python
sent
```

```python
## Define the vocubulary size 
voc_size=10000
```

```python
## One Hot Representation 
one_hot_repr =[]

# Empty list to store one-hot results
one_hot_repr = []

# Loop through each word
for word in sent:

    # Convert word into one-hot representation
    result = one_hot(word, voc_size)

    # Store the result
    one_hot_repr.append(result)
```
