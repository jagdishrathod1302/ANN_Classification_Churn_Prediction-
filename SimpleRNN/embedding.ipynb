```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent=[  'the glass of milk',
     'the glass of juice',
     'the cup of tea',
    'I am a good boy',
     'I am a good developer',
     'understand the meaning of words',
     'your videos are good',]
```

```python
sent
```

```python
## Define the vocabulary size
voc_size=10000
```

```python
sent
```

```python
### One Hot Representation
one_hot_repr=[one_hot(words,voc_size)for words in sent]
one_hot_repr
```

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
##basic the our RNN model needs sequences in a batch to have the same length 
## pad_sequence() : add padding or truncate sequence to a fixed lenght.
## paddig='pre': add zeros at the begining of shorter sequences.
## maxlen = sent_length: make every sequence exactly 8 element long  

sent_length = 8

embbeding_docs = pad_sequences(one_hot_repr,padding='pre',maxlen=sent_length)
print(embbeding_docs)
```

```python
dim = 10
```

```python
model = Sequential()

model.add(Embedding(voc_size,dim,input_length=sent_length))
model.compile("adam",'mse')
```

```python
model.summary()
```

```python
#Embedding layer, but Keras hasn't established the model's complete input shape yet. thats why do below step

#Import Sequential to create a neural network  model 
from tensorflow.keras.models import Sequential

#Import Embedding: to covert the wordID into dense vectotrs
#Impot Input : to define the shape of input data
from tensorflow.keras.layers import Embedding,Input

#create sequential empty neural network model 
model = Sequential()

#Define the input shape
# Each sentence contains a fixed number of word IDs
# sent_length = number of word IDs in each sentence
model.add(Input(shape=(sent_length,)))

# Add the Embedding layer
model.add(
    Embedding(
        input_dim=voc_size, # Total number of words in the vocabulary
        output_dim=dim # Number of values in each word vector
    )
)

# Configure the model for training
# adam = optimizer that updates model weights
# mse = Mean Squared Error, a loss function
model.compile(optimizer='adam',loss='mse')

# Display the model architecture, output shapes, and parameters
model.summary()
```

```python
# Generate output vectors for the input sentences 
prediction = model.predict(embbeding_docs)
```
